## 10. Feature Engineering

Creates new useful columns from existing data to make analysis easier and help answer business questions.

```python
# Here we'll create a few useful new columns from the existing data, rather than creating lots of unnecessary features.
```

```python
# 1.Create Order Month
# Extracts the year and month from order_date so we can analyze monthly sales and profit trends.

df["order_month"] = df["order_date"].dt.to_period("M").astype(str)
```

```python
df[["order_date", "order_month"]].head()
```

```python
# Create Delivery Delay
# delivery_days = actual delivery time
# estimated_delivery_days = expected delivery time
# Calculates the difference between actual and estimated delivery time.
# we can analyze delivery performance and customer satisfaction.

df["delivery_delay"] = (
    df["delivery_days"] - df["estimated_delivery_days"]
)
```

```python
df[[
    "delivery_days",
    "estimated_delivery_days",
    "delivery_delay"
]].head()
```

## 11. Outlier Detection

```python
# Outlier Detection
financial_columns = [
    "gross_sales",
    "discount_amount",
    "net_sales",
    "product_cost",
    "profit",
    "customer_lifetime_value"
]

for col in financial_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    
    print(f"{col}: {outliers} outliers")
```

## 12. Exploratory Data Analysis (EDA)

EDA means exploring the cleaned dataset to find patterns, trends, relationships, and useful business insights.

```python
# 12.1 Monthly Sales and Profit Trend
# How do sales and profit change over time?
# How are the company's sales and profit performing month by month?

monthly_sales = (
    df.groupby("order_month")[["net_sales", "profit"]]
    .sum()
    .reset_index()
)

monthly_sales.head()
```

```python
# 12.2 — Sales by Customer Segment
# Groups the orders according to customer segment and calculates the total Net Sales generated by each segment.
# Which customer segments contribute the most to total sales?

segment_sales = (
    df.groupby("customer_segment")["net_sales"]
    .sum()
    .reset_index()
    .sort_values("net_sales", ascending=False)
)

segment_sales
```

```python
# 12.3 — Sales by Region
# Groups the orders by region and calculates the total Net Sales for each region.
# Which regions generate the highest and lowest total sales?

region_sales = (
    df.groupby("region")["net_sales"]
    .sum()
    .reset_index()
    .sort_values("net_sales", ascending=False)
)

region_sales
```

```python
# 12.4 — Sales by Customer Type
# Groups the orders by customer type and calculates the total Net Sales for each type.
# Which customer type contributes the most to total sales?

customer_type_sales = (
    df.groupby("customer_type")["net_sales"]
    .sum()
    .reset_index()
    .sort_values("net_sales", ascending=False)
)

customer_type_sales
```

```python
# 12.5 — Sales by Gender
# Groups the orders by gender and calculates the total Net Sales for each gender.
# How is total sales distributed across genders?

gender_sales = (
    df.groupby("gender")["net_sales"]
    .sum()
    .reset_index()
    .sort_values("net_sales", ascending=False)
)

gender_sales
```

```python
# 12.6 — Sales by Payment Method
# Groups orders according to their payment method and calculates total Net Sales for each method.
# Which payment methods are associated with the highest total sales?

payment_sales = (
    df.groupby("payment_method")["net_sales"]
    .sum()
    .reset_index()
    .sort_values("net_sales", ascending=False)
)

payment_sales
```

```python
# 12.7 — Sales by Sales Channel
# Groups orders according to the sales channel and calculates total Net Sales for each channel.
# Which sales channel generates the most total sales?

channel_sales = (
    df.groupby("sales_channel")["net_sales"]
    .sum()
    .reset_index()
    .sort_values("net_sales", ascending=False)
)

channel_sales
```

```python
# 12.8 — Profit by Customer Segment
# Groups orders by customer segment and calculates the total Profit generated by each segment.
# Which customer segment generates the most total profit?

segment_profit = (
    df.groupby("customer_segment")["profit"]
    .sum()
    .reset_index()
    .sort_values("profit", ascending=False)
)

segment_profit
```

```python
# 12.9 — Profit by Region
# Compares total profit generated by each geographical region
# Which region generates the highest total profit?

region_profit = (
    df.groupby("region")["profit"]
    .sum()
    .reset_index()
    .sort_values("profit", ascending=False)
)

region_profit
```

```python
# 12.10 — Profit by Sales Channel
# Compares the total profit generated through each sales channel.
# Which sales channel generates the highest total profit?

channel_profit = (
    df.groupby("sales_channel")["profit"]
    .sum()
    .reset_index()
    .sort_values("profit", ascending=False)
)

channel_profit
```

```python
# 12.11 — Repeat vs New Customer Performance
# Compares sales and profit from repeat customers versus non-repeat customers.
# Do repeat customers contribute more sales and profit?

repeat_customer_performance = (
    df.groupby("is_repeat_customer")[["net_sales", "profit"]]
    .sum()
    .reset_index()
)

repeat_customer_performance
```

```python
# 12.12 — Discount vs Sales & Profit
# Compares sales and profit for orders where a discount was applied versus orders without a discount.
# How do sales and profit differ between discounted and non-discounted orders?

df["discount_status"] = df["discount_amount"].apply(
    lambda x: "Discount Applied" if x > 0 else "No Discount"
)

discount_performance = (
    df.groupby("discount_status")[["net_sales", "profit"]]
    .sum()
    .reset_index()
)

discount_performance
```

```python
# 12.13 — Delivery Status vs Sales & Profit
# Compares sales and profit across different delivery statuses.
# Which delivery status contributes the most sales and profit?

delivery_performance = (
    df.groupby("delivery_status")[["net_sales", "profit"]]
    .sum()
    .reset_index()
    .sort_values("net_sales", ascending=False)
)

delivery_performance
```

```python
# 12.14 — Marketing Channel Performance
# Compares sales and profit generated through different marketing channels.
# Which marketing channel generates the highest total sales and profit?

# Marketing Channel Performance

marketing_performance = (
    df.groupby("marketing_channel")[["net_sales", "profit"]]
    .sum()
    .reset_index()
    .sort_values("net_sales", ascending=False)
)

marketing_performance
```

```python
# 12.15 — Return Analysis
# Examines returned vs non-returned orders and identifies the main reasons for returns.

# Return Status Analysis

return_analysis = (
    df.groupby("return_status")[["net_sales", "profit"]]
    .sum()
    .reset_index()
    .sort_values("net_sales", ascending=False)
)

return_analysis
```

```python
# Return Reason Analysis
# What are the most common reasons for product returns?

return_reason_analysis = (
    df[df["return_status"] == "Returned"]["return_reason"]
    .value_counts()
    .reset_index()
)

return_reason_analysis.columns = ["return_reason", "return_count"]

return_reason_analysis
```
